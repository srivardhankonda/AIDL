```python
import os
import zipfile
from pathlib import Path
import torch
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

uploaded = files.upload()

zip_files = [f for f in uploaded if f.lower().endswith(".zip")]

if not zip_files:
    raise ValueError("Please upload a ZIP dataset.")

ZIP_PATH = zip_files[0]
EXTRACT_DIR = "/content/waste_dataset"

os.makedirs(EXTRACT_DIR, exist_ok=True)

with zipfile.ZipFile(ZIP_PATH, "r") as archive:
    bad_file = archive.testzip()

    if bad_file:
        raise ValueError(f"Corrupted ZIP file: {bad_file}")

    archive.extractall(EXTRACT_DIR)

print("Dataset extracted successfully!")
print("Dataset location:", EXTRACT_DIR)
```
